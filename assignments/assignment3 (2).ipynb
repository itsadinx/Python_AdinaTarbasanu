### Loading the news data

````python
news = pd.read_csv("news.csv", sep=";")
display(news)
````

````python
#a)
import json

def classify_news(headline, model):
    """Classify a financial news headline by topic and sentiment.
    Returns a dict with 'topic' and 'sentiment' keys."""
    prompt = f"""Classify the following financial news headline.
Return a JSON object with exactly two keys:
- "topic": one of "earnings", "mergers", "regulation", "macroeconomics"
- "sentiment": one of "positive", "negative", "neutral"

Do not include any other text, markdown formatting, or code blocks. Return only the JSON object.

Headline: {headline}"""
    result = ask_llm(prompt, model)
    # Clean in case model wraps in markdown fences (as shown in L5_3_text_analysis)
    cleaned = result.strip().removeprefix("```json").removeprefix("```").removesuffix("```").strip()
    return json.loads(cleaned)
````

````python
#b)
classifications = news["headline"].apply(lambda h: classify_news(h, MODEL_LARGE))

news_classified = news.copy()
news_classified["topic"] = classifications.apply(lambda x: x["topic"])
news_classified["sentiment"] = classifications.apply(lambda x: x["sentiment"])

display(news_classified)
````
